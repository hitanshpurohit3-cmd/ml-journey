```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,11)
x
y = x**2
y
```

```python
#plot makes a line graph
#a plot has two lines one vertical(y) and one horizontal(x)
plt.plot(x , y )
```

```python
plt.plot((x**2),(y)**3)
plt.title("Hello") # giving title to the graph
plt.xlabel("Quantity",loc='left')#adding label to the x-axis
plt.ylabel("Total Bill")#adding label to the y-axis
```

```python
#here in this we have to provide row and axis
#subplots either takes one or three positional arguments two are not acceptable
# in case of three positional argumnets the first two re row,column and the third one is plot selection
plt.subplot(2,2,1)
plt.subplot(2,2,4)
```

```python
#we can write these in any manner
plt.subplot(2,2,1)
plt.subplot(2,2,3)
plt.subplot(2,2,4)
plt.subplot(2,2,2)
```

```python
# suppose in the 4th plot i want to enter some data
#we have to write the data values below we address the subplots
plt.subplot(2,2,1)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot((y)**2,(x)**3)
plt.subplot(2,2,3)
plt.plot(y,x)
```

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(x,y)
```

```python
#Creating multiple axis in the same figure
fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)
axis2 = fig.add_axes([0.1,0.1,0.7,0.7])
axis2.plot(x,y)
```

```python
#creating multiple axes in single figure(canvas)
fig = plt.figure()
axis1=fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)
axis2=fig.add_axes([0.3,0.5,0.7,0.7])
axis2.plot(x,y)
```

Matplotlib allows the aspect ratio, DPI and figure size to be specified when the Figure object is created. You can use the figsize and dpi keyword arguments.
figsize is a tuple of the width and height of the figure in inches dpi is the dots-per-inch(pixel per inch).

```python
#creating multiple axes in single figure(canvas)
fig = plt.figure(figsize=(6,8),dpi= 100)#figsize(h,w)
axis1=fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)
axis2=fig.add_axes([0.3,0.1,0.7,0.7])
axis2.plot(x,y)
```

```python
#scatterplot used to make regression algorithm
plt.scatter(x,y)
```

```python
#histogram
#in histogram we have to provide only x and y automatically works with respect to points
from random import sample
data =  sample(range(1,1000),100)
plt.hist(data)
```

```python
#Box plot
#it also has only x needed y is not needed
plt.boxplot(x)
```

```python
#Saving image
fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig("basicplot.png")
```
