```python
import seaborn as sns
import matplotlib.pyplot as plt
```

Categorical Data Plots
Now let's discuss using seaborn to plot categorical data! There are a few main plot types for this:
* boxplot
* violinpot
* stripplot
* swarmplot
* barplot
* countplot
Let's go through examples of each!

```python
dataset = sns.load_dataset('tips')
```

```python
dataset
```

```python
dataset.columns
```

Countplot

```python
sns.countplot(dataset)
```

```python
sns.countplot(dataset['sex'])
```

```python
sns.countplot(x= dataset['sex'])
```

```python
#to get sexwise smoker
sns.countplot(x=dataset['sex'], hue = dataset['smoker'])
```

Bar plot (good to use atleast two parameters).
in bar plot:
* x axis represents the categorical datatype or qualitative variable
* y axis represents the distributive datatype or quantitative variable

```python
sns.barplot(x = dataset['sex'], y = dataset['smoker'])
#here we can see it will definitely not give any error while making the barplot of distributive to distributive variable but it is wrong conceptually
```

```python
#this is the correct plotting of barplot that is categorical to distributive variable
sns.barplot(x = dataset['sex'],y = dataset['tip'])
```

You can change the estimator object to your own function, that converts a vector to a scalar
* Mean(default): np.mean
* Median: np.median
* Sum: np.sum
* Count: len(to count the number of observations)
* Standard Deviation: np.std
* Minimum: np.min
* Maximum: np.max

```python
import numpy as np
sns.barplot(x= dataset['sex'], y= dataset['tip'],estimator = np.sum)
```

```python
sns.barplot(x= dataset['sex'],y = dataset['total_bill'],estimator = np.sum)
```

Boxplot(very important in statistics)


```python
sns.boxplot(x ='tip', y ='day',data = dataset)# x-axis is distributive variable, y-axis is categorical variable
```

```python
sns.boxplot(x='tip',y='day',data = dataset, palette ='rainbow')
```

BoxPlot and violinPlot are used to show the distribution of categorical data. A box plot(or box-and-whisker plot) shows the distribution of quantitative data in a way that facilitates comparison between variables or across levels of a categorical variable. The box shows the quaritles of the dataset while the whiskers extend to show the rest of the distribution, except for points that are determined t be "outliers" using a method that is a function of the inter-qurtile range.

ViolinPlot

```python
sns.violinplot(x ='day', y = 'total_bill' ,data = dataset)
```

## StripPlot and SwarmPlot
The stripplot will draw a scatterplot where one variable is categorical. A strip plot can be drawn on its own, but it is also a good complement to a box or violin plot in cases where you want to show all obersvations along with some representation of the underlying distribution.

The swarmplot is similar to stripplot(), but the points are adjusted (only along the categorical axis) so that they don't overlap. This gives a better representation of the distribution of the distribution of values, although it doesn't scalse as well to large numbers of observations(both in terms of the ability to show all the points and in terms of the computation needed to arrange them).

```python
sns.stripplot(x ='day', y='total_bill', data=dataset)
```

```python
sns.swarmplot(x='day', y='total_bill', data=dataset)
```

```python
#combining a swarmplot and a violin plot
sns.violinplot(x ='tip', y='day',data=dataset)
sns.swarmplot(x ='tip', y='day',data=dataset)
```

```python
#combining a swarmplot and a violin plot
sns.violinplot(y ='tip', x='day',data=dataset)
sns.swarmplot(y ='tip', x='day',data=dataset)
```
