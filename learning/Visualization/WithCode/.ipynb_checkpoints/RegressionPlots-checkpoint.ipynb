```python
import seaborn as sns
import matplotlib.pyplot as plt
```

## Implot()
regression plot is used in machine learning so we will not fully discuss what, how and where it is used but we will see how it is created and what are the values we can put inside it

```python
tips = sns.load_dataset('tips')
tips
```

```python
sns.lmplot(x='total_bill', y='tip', data=tips)#y-axis holds what to predict and x-axis holds whose through you want to predict
```

```python
sns.lmplot(x='total_bill',y='tip', hue = 'sex', data = tips)
```

```python
sns.lmplot(x='total_bill',y='tip',hue='sex',data=tips,palette='coolwarm')
```

```python
sns.lmplot(x='total_bill', y='tip', data=tips , hue='sex', palette='coolwarm', markers=['o','v'])
```

```python
sns.lmplot(x='total_bill', y='tip', data=tips, hue='sex', palette='rainbow', markers=['o','v'], scatter_kws={'s':200})#by default is 50 size
```
