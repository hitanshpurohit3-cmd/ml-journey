```python
import numpy as np
import pandas as pd
```

```python
dataset = pd.read_csv(r"Countries.csv")
```

```python
dataset.head(5)
```

```python
dataset.columns
```

```python
#which country has the highest population
dataset[dataset["population"] == dataset["population"].max()]["country"]
```

```python
#what is the capital of the country with highest population
dataset[dataset["population"] == dataset["population"].max()][["country","capital_city"]]
```

```python
#which country has the least population
dataset[dataset["population"] == dataset["population"].min()][["country","population"]]
```

```python
#What is the capital of the country with the least population
dataset[dataset["population"] == dataset["population"].min()][["country","capital_city","population"]]
```

```python
#give me the top 5 countries with highest democratic score
dataset.sort_values(by='democracy_score',ascending = False).head(5)[["country","democracy_score"]]
```

```python
#how many total regions are there
dataset['region'].value_counts().count()
```

```python
# how many countries lie in Eastern Europe region
dataset[dataset['region'] == 'Eastern Europe']['country']
```

```python
#Who is the political leader of the second highest populated country method a
dataset.sort_values(by='population', ascending = False).iloc[1]['political_leader']
```

```python
#who is the political leader of the second highest populated country method b
dataset[dataset["population"] == dataset["population"].nlargest(2).iloc[1]]
```

```python
#how many countries are there whose political leaders are unknown
dataset[dataset['political_leader'].isna()]['country'].count()
```

```python
#how many country have republica in their full name
count = 0
def counting(txt):
    global count
    if 'republic' in txt.lower():
        count +=1
    return txt
dataset["country_long"] = dataset['country_long'].apply(counting)
print(count)
```

```python
#which country in africa has highest population
africa_df = dataset[dataset['continent'] == 'Africa']
```
