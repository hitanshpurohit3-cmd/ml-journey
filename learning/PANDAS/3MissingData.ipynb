```python
import numpy as np
import pandas as pd
```

```python
data = {
    'A': [1,2,np.nan,4,5],
    'B': [np.nan, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan ,np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
df2 = pd.DataFrame(data)
```

```python
df
```

```python
#gives boolean mapping of whether a valus is null or not
df.isna()
```

```python
#Gives columnwise sum of null values
df.isna().sum()
```

```python
# gives whether a column has null values or not
df.isna().any()
```

```python
df
```

```python
# drops all rows with null values
df2.dropna()
```

```python
df
```

```python
#thresh acts as a threshold value i.e, here all rows atleast 3 non null values won't be affected and all below 3 non null vlues will get deleted
df.dropna(thresh=3)
```

```python
#fill all null values with the given parameterized value in the fillna attribute
df.fillna(0)
```

```python
# to fill different values in place of each missing value in different columns
values = {'A':0,'B':100,'C':300,'D':400}
df.fillna(value = values)
```

```python
#to fill missing values with mean
df.fillna(df.mean())
```
